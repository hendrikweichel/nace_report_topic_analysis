```python
def get_model(ckpt_path, num_layers=2):
    # load checkpoint
    config = AutoConfig.from_pretrained(ckpt_path)

    # 2. Build a model from config (bare BertForSequenceClassification)
    model = AutoModelForSequenceClassification.from_config(config)

    # 3. Rebuild the SAME classifier architecture as in training
    hidden = getattr(config, "custom_hidden", 512)  # fallback if not in config
    num_layers = getattr(config, "custom_num_layers",num_layers)

    layers = []
    for i in range(num_layers):
        in_dim = config.hidden_size if i == 0 else hidden
        layers.append(nn.Linear(in_dim, hidden))
        layers.append(nn.GELU())
        layers.append(nn.Dropout(0.2))
    layers.append(nn.Linear(hidden, config.num_labels))
    model.classifier = nn.Sequential(*layers)

    # 4. Load weights from model.safetensors
    state_dict = load_file(os.path.join(ckpt_path, "model.safetensors"))
    model.load_state_dict(state_dict, strict=True)  # will fail loudly if mismatch

    # 5. Inference mode
    model.eval()
    return model
```

```python
from IPython.display import HTML

def show_token_attributions(tokens, scores, cmap='RdYlGn'):
    """
    Display tokens with background color based on attribution score.
    """
    import matplotlib.cm as cm
    import matplotlib.colors as colors

    norm = colors.Normalize(vmin=min(scores), vmax=max(scores))
    colormap = cm.get_cmap(cmap)

    html_tokens = []
    for t, s in zip(tokens, scores):
        rgba = colormap(norm(s))
        color = f"rgba({int(rgba[0]*255)}, {int(rgba[1]*255)}, {int(rgba[2]*255)}, 0.6)"
        html_tokens.append(
            f"<span style='background-color:{color}; padding:2px; margin:2px; border-radius:3px;'>{t}</span>"
        )

    return HTML(" ".join(html_tokens))
```

```python
from captum.attr import IntegratedGradients
import torch

inputs = tokenizer(chunk, return_tensors="pt", truncation=True)
input_ids = inputs["input_ids"]
attention_mask = inputs["attention_mask"]

# Forward pass to choose class
with torch.no_grad():
    outputs = model(**inputs)
    probs = torch.softmax(outputs.logits, dim=-1).squeeze(0)

predicted_class = torch.argmax(probs).item()

# IG setup
def forward_func(embeddings):
    out = model(inputs_embeds=embeddings, attention_mask=attention_mask)
    return out.logits[:, predicted_class]

ig = IntegratedGradients(forward_func)

# Embeddings
embeddings = model.get_input_embeddings()(input_ids)

# Baseline = zero embeddings
baseline = torch.zeros_like(embeddings)

# Attribution
attributions, delta = ig.attribute(
    embeddings,
    baselines=baseline,
    return_convergence_delta=True
)

# Token-level attribution (sum embedding dims)
token_attributions = attributions.sum(dim=-1).squeeze(0)
tokens = tokenizer.convert_ids_to_tokens(input_ids.squeeze(0))

token_importance = list(zip(tokens, token_attributions.tolist()))

# Your original label scores
label_scores = {
    model.config.id2label[i]: probs[i].item()
    for i in range(len(probs))
}
label_scores = dict(sorted(label_scores.items(), key=lambda x: x[1], reverse=True))

print("Prediction:", label_scores)
print("Integrated Gradients token importance:")
for tok, score in token_importance:
    print(f"{tok:15s} {score:+.4f}")
```

```python
def predict_and_explain(chunk, model, tokenizer, device="cpu"):
    # ---- Standard prediction ----
    inputs = tokenizer(chunk, return_tensors="pt", truncation=True)
    inputs = {k: v.to(device) for k, v in inputs.items()}
    input_ids = inputs["input_ids"]
    attention_mask = inputs["attention_mask"]

    with torch.no_grad():
        outputs = model(**inputs)
        probs = torch.softmax(outputs.logits, dim=-1).squeeze(0)

    predicted_class = torch.argmax(probs).item()

    # ---- Integrated Gradients ----
    # 1) Get embeddings (this uses input_ids as Long => OK)
    input_ids_long = input_ids.long()  # just to be explicit
    embeddings = model.get_input_embeddings()(input_ids_long)

    # 2) Define forward function: IMPORTANT -> use inputs_embeds=...
    def forward_func(emb):
        out = model(
            inputs_embeds=emb,             # <--- not input_ids=emb
            attention_mask=attention_mask
        )
        return out.logits[:, predicted_class]

    ig = IntegratedGradients(forward_func)

    # 3) Baseline: zero embeddings
    baseline = torch.zeros_like(embeddings)

    # 4) Attribution
    attributions, delta = ig.attribute(
        embeddings,
        baselines=baseline,
        return_convergence_delta=True
    )

    # 5) Token-level attribution (sum embedding dims)
    token_attributions = attributions.sum(dim=-1).squeeze(0)
    tokens = tokenizer.convert_ids_to_tokens(input_ids.squeeze(0))

    token_importance = list(zip(tokens, token_attributions.tolist()))

    label_scores = {
            model.config.id2label[i]: probs[i].item()
            for i in range(len(probs))
        }

    return label_scores, token_importance, delta.item()
```

```python
def plot_token_importance(token_importance, title="Token Importance (Integrated Gradients)", predicted_class: str = "", right_class: str = ""):
    tokens = [t for t, s in token_importance]
    scores = np.array([s for t, s in token_importance])

    # Color map: positive (red), negative (blue)
    colors = ['red' if s > 0 else 'blue' for s in scores]

    plt.figure(figsize=(10, max(4, len(tokens) * 0.2)))
    y_pos = np.arange(len(tokens))

    plt.barh(y_pos, scores, color=colors)
    plt.yticks(y_pos, tokens, fontsize=10)
    plt.xlabel("Attribution Score")
    
    if predicted_class in "ABCDEFGHIJKLMNOPQRSTUVW" and predicted_class != "":
        title += ", predicted class: " +  predicted_class + ": " + get_description(predicted_class)
    else:
        title += ", predicted class: " +  predicted_class

    if right_class in "ABCDEFGHIJKLMNOPQRSTUVW" and right_class != "":
        title += "\nright class: " +  right_class + ": " + get_description(right_class)
    else:
        title += "\nright class: " +  right_class
        
    plt.title(title)
    plt.gca().invert_yaxis()  # token order top-to-bottom

    plt.tight_layout()
    plt.show()
```

```python
def classify_and_explain(chunk, model, tokenizer, device="cpu", right_class=""):
    """
    Classifies a chunk using the given model, performs Integrated Gradients (IG),
    and plots the token importance.

    Args:
        chunk (str): The text chunk to classify.
        model (torch.nn.Module): The pre-trained model for classification.
        tokenizer (transformers.PreTrainedTokenizer): The tokenizer for the model.
        device (str): The device to run the model on ("cpu" or "cuda").

    Returns:
        dict: The label scores from the classification.
        list: The token importance scores.
    """
    # Classify and explain using the predict_and_explain function
    label_scores, token_importance, delta = predict_and_explain(chunk, model, tokenizer, device)
    sorted_items = sorted(label_scores.items(), key=lambda x: x[1], reverse=True)
    for key, value in sorted_items:
        print(f"{key}: {value:.6f}")
        
    # Get the predicted class for the title of the plot
    predicted_class = max(label_scores, key=label_scores.get)

    # Plot the token importance
    plot_token_importance(
        token_importance,
        title="Token Importance (Integrated Gradients)",
        predicted_class=predicted_class,
        right_class=right_class
    )
    return label_scores, token_importance
```
