```python
from langchain_ollama import OllamaLLM

def query_ollama(prompt: str, model: str = "llama3.2:1b") -> str:
    """
    Simple function to call an Ollama model through LangChain.

    Args:
        prompt (str): The input text prompt for the model.
        model (str): The name of the Ollama model to use (default: "llama3").

    Returns:
        str: The model's generated response.
    """
    llm = OllamaLLM(model=model)
    response = llm.invoke(prompt)
    return response
```

```python
descriptions = []

for csv in tqdm.tqdm(csvs): 

    df = pd.read_csv(csv)

    pages = "\n".join([f.replace("/spacelining", "") for f in df.loc[:10, "text"]])

    prompt = f"""
    I will give you a set of pages extracted from the same annual report.  
    Your task is to read all pages together and produce a concise, structured summary of the company's business model **based only on the information in these pages**, i.e. the sector the company works in.

    Please extract:
    1. The core business model in 4-6 sentences.
    2. Main revenue streams.
    3. Key products/services.
    4. Target customer segments and geographic focus.
    5. Strategic priorities mentioned in the text.

    If information is missing or unclear, explicitly state that it is not provided in the given pages.

    Here are the pages:
    {pages}
    """

    summary = query_ollama(prompt, "llama3:8b")

    print(summary)

    descriptions.append({"company": os.path.basename(csv)[:-4], "summmary": summary})
```
