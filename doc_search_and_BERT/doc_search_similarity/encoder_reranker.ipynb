```python
from sentence_transformers import CrossEncoder
import pandas as pd
```

```python
df = pd.read_csv('/Users/hendrikweichel/projects/NaceCodeClassification/nace_report_topic_analysis_3/results/similarity_search_descriptions/title_and_text__all_reports_2/Allfunds Group plc1.csv')
df
```

```python
reranker = CrossEncoder('cross-encoder/ms-marco-MiniLM-L-6-v2')
```

```python
# Query and retrieved docs
query = "What are the business activities, i.e. the sector or branche of the company? With what does the company make money?"
docs = df.loc[:5, "text"]

# Pair up query with each doc
pairs = [(query, doc) for doc in docs]

# Predict relevance scores
scores = reranker.predict(pairs)

# Sort and rerank
reranked = [doc for _, doc in sorted(zip(scores, docs), reverse=True)]

# Print the reranked documents with their scores
for score, doc in zip(sorted(scores, reverse=True), reranked):
    print(f"Score: {score:.2f}")
    print(f"Document: {doc}\n")
```

```python
for i in range(len(scores)): 
    print(i)
    print(scores[i])
    print(docs[i])
```

```python
from langchain.llms import Ollama

# Initialize the LLM
llm = Ollama(model="llama3:8b")

# Prepare the reranking prompt
prompt = f"""
You are a helpful assistant. Rerank the following documents based on their relevance to the query: "{query}".

Documents:
{[doc for doc in docs]}

Return the reranked documents in descending order of relevance.
"""

# Get the reranked documents
reranked_docs = llm(prompt)

# Print the reranked documents
print(reranked_docs)
```

```python
from langchain.llms import Ollama

# Initialize the LLM
llm = Ollama(model="llama3:8b")

# Prepare the reranking prompt
prompt = f"""
You are a helpful assistant. Return the document with the highest relevance for this query: "{query}".

Documents:
{"".join(['Document ID ' + str(i) + doc for i, doc in enumerate(docs)])}

Only return the Document ID with the highest relevance without any explanation.
"""

# Get the reranked documents
reranked_docs = llm(prompt)

# Print the reranked documents
print(reranked_docs)
```
