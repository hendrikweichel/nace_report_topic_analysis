```python
import matplotlib.pyplot as plt
import fitz  # PyMuPDF
import tqdm
import glob
```

```python
pdf_path = '/Users/hendrikweichel/projects/NaceCodeClassification/nace_report_topic_analysis_3/data/PDF_stoxx600/Zurich Insurance Group Ltd2.pdf'
```

```python
doc = fitz.open(pdf_path)
all_tabs = []

for i, page in tqdm.tqdm(enumerate(doc), total=len(doc), desc="Extracting tables"):
    # Detect tables on the current page
    tabs = page.find_tables()

    if tabs.tables:  # Check if there are any tables
        for j, table in enumerate(tabs.tables):
            # Convert table to markdown
            md = table.to_markdown()
            
            # Store useful metadata too
            all_tabs.append({
                "page_num": i + 1,
                "table_num": j + 1,
                "markdown": md
            })

# Example: write all to file
with open("tables.md", "w", encoding="utf-8") as f:
    for t in all_tabs:
        f.write(f"### Page {t['page_num']} - Table {t['table_num']}\n\n")
        f.write(t["markdown"])
        f.write("\n\n")
```
