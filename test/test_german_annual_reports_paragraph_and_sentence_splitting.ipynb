## Test retrieving the similarities for chunks in a pdf to the NACE Code

**Function:** pdf-> (chunk x code -> [-1,1])

**Parameters:** 

- pdf_path
- way of chunking the text (e.g. sentences, sliding window, or paragraphs)
- way of preprocessing (most is fixed for all reports)
    - similarity threshold of relevant chunks
    - length of irrelevant chunks

**Store analytics for each datapoint:**

- mean score for each class given a threshold

```python
# Parameters: 

threshold_min_chunk_len = 100
cos_threshold = 0.4
sentence_length = 3

result_path = f"../results/paragraph_and_sentence_len_{sentence_length}_min_chunk_len_{threshold_min_chunk_len}_cos_thresh_{cos_threshold}_stoxx"
```

```python
dataset_path = "../data/german_annual_reports"
dataset_path = "../data/PDF_stoxx600"
```

```python
nace_classes = pd.read_excel(os.path.join(dataset_path, "STOXX600_as_of_2025_03_13.xlsx"))
nace_classes.head()
```

```python
report_to_nace_class = nace_classes.set_index('Report').to_dict()["NACE"]
report_to_nace_class
```

```python
reports_path = glob.glob(os.path.join(dataset_path, "*.pdf"))
```

```python
def preprocess_report(pdf_path: str) -> List[str]:
    chunks = text_extraction.get_chunked_paragraphs(pdf_path, sentence_length)
    chunks = [text_extraction.preprocess_text(chunk) for chunk in chunks]
    return chunks
```

```python
res = test_base.test_similarities(reports_path, preprocess_report, threshold_min_chunk_len, cos_threshold, report_to_nace_class, result_path, level=4)
```
