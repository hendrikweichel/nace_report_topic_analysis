```python
pdfs = glob.glob(pdfs_path + "/*.pdf")
```

```python
def get_level_1_nace(x):
    x = float(x)
    
    if 1 <= x <= 3.22:
        return 'A'
    elif 5 <= x <= 9.9:
        return 'B'
    elif 10 <= x <= 33.20:
        return 'C'
    elif 35 <= x <= 35.30:
        return 'D'
    elif 36 <= x <= 39:
        return 'E' 
    elif 41 <= x <= 43.99:
        return 'F'
    elif 45 <= x <= 47.99:
        return 'G'
    elif 49 <= x <= 53.2:
        return 'H'
    elif 55 <= x <= 56.3:
        return 'I'
    elif 58 <= x <= 63.99:
        return 'J'
    elif 64 <= x <= 66.3:
        return 'K'
    elif 68 <= x <= 68.32:
        return 'L'
    elif 69 <= x <= 75:
        return 'M' 
    elif 77 <= x <= 82.99:
        return 'N'
    elif 84 <= x <= 84.3:
        return 'O'
    elif 85 <= x <= 85.6:
        return 'P'
    elif 86 <= x <= 88.99:
        return 'Q'
    elif 90 <= x <= 93.29:
        return 'R'
    elif 94 <= x <= 96.09:
        return 'S' 
    elif 97 <= x <= 98.2:
        return 'T'
    elif x == 99:
        return 'U'
    else:
        return None

```

```python
full_ds = full_ds[full_ds["Entity NACE Code"] != "@NA"]
```

```python
full_ds["NACE_Letter"] = full_ds["Entity NACE Code"].apply(get_level_1_nace)
```

# Copy JSON to TXT

```python
import json
```

```python
jsons_path = "projects/nace_classification/nace_report_topic_analysis/data/datasets/reports_subset_from_full_data_3/JSONs"
jsons_paths = glob.glob(jsons_path + "/*.json")
```

```python
for i in range(len(jsons_paths)): 
    with open(jsons_paths[i], "r") as f: 
        json_report = json.load(f)

    text = "\n".join([page["markdown"] for page in json_report["pages"]])
    text_path = jsons_paths[i].replace("JSONs", "TXTs").replace(".json", ".txt")

    with open(text_path, "w") as f:
        f.write(text)
```
