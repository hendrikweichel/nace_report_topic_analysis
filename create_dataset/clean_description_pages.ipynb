```python
import pandas as pd
import os
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
```

```python
system_prompt_format = "You are a NLP assistant that cleans text that was extracted via OCR. Only return text that is in the output!"
prompt_format = "The following text was extracted via OCR. Please only extract the paragraph in it that describes the companies busines model. No Headers, tables, or text that does not contain information about the busines model.\n\nText: {text}"
```

```python
def clean_description(
        text: str,
        temperature: float = 0.1
): 

    # Initialize LLM
    llm = ChatOpenAI(
        model="gpt-4o-mini",
        temperature=temperature
    )

    # Prompt
    prompt = ChatPromptTemplate.from_messages([
        ("system", system_prompt_format),
        ("human", prompt_format)
    ])
    
    # Chain
    chain = prompt | llm

    # inputs
    input = {
        "text": text,
        }

    formatted_prompt = prompt.invoke(input)

    # Run
    response = chain.invoke(input)

    return formatted_prompt, response.content
```

```python
ds_2_desc  = pd.read_csv("projects/nace_classification/nace_report_topic_analysis/data/datasets/reports_subset_from_full_data_2/reports_subset_from_full_data_2_overview_with_description.csv")
```

```python
ds_2_desc[pd.notna(ds_2_desc["Description"])].groupby("NACE_letter").size()
```

```python
ds_2_desc.loc[pd.notna(ds_2_desc["Description"]), "Description_clean"] = ds_2_desc.loc[pd.notna(ds_2_desc["Description"]), "Description"].apply(clean_description)
```
