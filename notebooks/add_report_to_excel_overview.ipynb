```python
import pandas as pd
import glob
import os
from fuzzywuzzy import fuzz
import numpy as np
```

```python
overview_path = "/Users/hendrikweichel/projects/NaceCodeClassification/nace_report_topic_analysis_3/data/PDF_stoxx600/STOXX600_as_of_2025_03_13.xlsx"
```

```python
df_overview = pd.read_excel(overview_path)
df_overview
```

```python
dataset_path_texts = "../data/PDF_stoxx600/"
reports_path = glob.glob(os.path.join(dataset_path_texts, "*.pdf"))
reports_path
```

```python
report_names = [os.path.basename(report).replace(".pdf", "") for report in reports_path]
report_names
```

```python
for i, row in df_overview.iterrows(): 
    if pd.notna(row["Report"]): 
        continue

    sims = report_names[np.argmax([fuzz.partial_ratio(name, row["Name"]) for name in report_names])]
    max_val = np.max([fuzz.ratio(name, row["Name"]) for name in report_names])
    if max_val > 70: 
        print(sims)
        print(row["Name"])
        print(max_val)
        print()
```
