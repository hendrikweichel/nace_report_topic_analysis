```python
import glob
import pandas as pd
from pypdf import PdfReader
import os
```

```python
reports_path = "/Users/hendrikweichel/projects/NaceCodeClassification/PDFs_selected/*"
reports_paths = glob.glob(reports_path)
len(reports_paths)
```

```python
for path in reports_paths:
    reader = PdfReader(path)

    num_pages = len(reader.pages)
    print("Number of pages:", num_pages)

    if num_pages < 40: 
        if os.path.exists(path):
            os.remove(path)
            print("File deleted.")
        else:
            print("File does not exist.")
```
